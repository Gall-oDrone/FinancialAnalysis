```python
from datetime import datetime, timedelta

def calculate_datetime_from_ago_string(ago_string):
    # Split the string by spaces
    parts = ago_string.split()
    
    # Extract the value and unit from the string
    value = int(parts[0])
    unit = parts[1].lower()  # Convert unit to lowercase for easier comparison
    
    # Determine the timedelta to subtract based on the unit
    if unit.endswith('s'):  # Check if the unit ends with 's' (plural)
        unit = unit[:-1]  # Remove the 's' to get the singular form
        
    # Map units to corresponding timedelta function
    unit_to_timedelta = {
        'second': timedelta(seconds=1),
        'minute': timedelta(minutes=1),
        'hour': timedelta(hours=1),
        'day': timedelta(days=1),
        'yesterday': timedelta(days=1),
        'days': timedelta(days=1),
    }
    
    # Calculate the timedelta based on the unit
    if unit in unit_to_timedelta:
        timedelta_to_subtract = unit_to_timedelta[unit] * value
        # Calculate the datetime and timestamp
        current_datetime = datetime.now()
        calculated_datetime = current_datetime - timedelta_to_subtract
        calculated_timestamp = int(calculated_datetime.timestamp())
        return calculated_datetime, calculated_timestamp
    else:
        raise ValueError("Invalid time unit: {}".format(unit))
```

```python
import pandas as pd

class DataFrameStore:
    def __init__(self, header=["id", "source", "headline", "href", "summary", "content", "author","minsread","datetime"]):
        self._data_frame = None
        self.header = header
        self.row_index = None

    @property
    def data_frame(self):
        return self._data_frame

    @data_frame.setter
    def data_frame(self, df):
        if df is not None and not isinstance(df, pd.DataFrame):
            raise ValueError("DataFrame must be a pandas DataFrame object")
        self._data_frame = df

    @property
    def data_frame_header(self):
        return self.header
    
    def create_data_frame(self):
        self.data_frame = pd.DataFrame(columns=self.header)

    def update_data_frame(self, data):
        # Check if DataFrame exists
        if self.data_frame is None:
            self.create_data_frame()
            print("DataFrame is not initialized. Creating a DataFrame.")

        # Check if the number of columns matches the length of the data
        if len(self.header) != len(data):
            raise ValueError("Number of columns does not match the length of the data.")

        # Map dictionary keys to DataFrame headers and assign values
        mapped_data = {}
        for header in self.header:
            mapped_data[header] = data.get(header, None)

        # Convert mapped data to DataFrame and concatenate with existing DataFrame
        new_data = pd.DataFrame(mapped_data, index=[0])
        new_data = pd.DataFrame(new_data)

        # Concatenate new data with existing DataFrame
        self.data_frame = pd.concat([self.data_frame, new_data], ignore_index=True)
```

```python
class NewsScraper:
    def __init__(self, debug=False, topics=["crypto"]):
        # Initialize the Selenium WebDriver with default options
        self.debug = debug
        self.set_driver_options()
        self.topics = topics
        
        # Check for system chromedriver (Docker environment)
        import os
        import shutil
        chromedriver_path = os.getenv('CHROMEDRIVER_PATH') or shutil.which('chromedriver')
        
        if chromedriver_path and os.path.exists(chromedriver_path):
            print(f"Using system chromedriver at: {chromedriver_path}")
            service = ChromeService(chromedriver_path)
        else:
            print("Using webdriver_manager to get chromedriver")
            service = ChromeService(ChromeDriverManager().install())
        
        # Set Chrome binary if specified (for Docker).
        # Prefer stable Chrome: chromium can exit immediately with the matching
        # chromedriver ("session not created: Chrome instance exited").
        chrome_bin = (
            os.getenv('CHROME_BIN')
            or shutil.which('google-chrome-stable')
            or shutil.which('google-chrome')
            or shutil.which('chromium')
        )
        if chrome_bin:
            self.options.binary_location = chrome_bin
            print(f"Using Chrome binary at: {chrome_bin}")
        
        self.driver = webdriver.Chrome(service=service, options=self.options)
        # Set page load timeout to prevent hanging
        self.driver.set_page_load_timeout(30)
        self.df_store = DataFrameStore()


    def set_driver_options(self):
        # Set desired options for the WebDriver
        import os
        self.options = webdriver.ChromeOptions()
        self.options.add_argument("--user-agent=Mozilla/5.0 (Macintosh; Intel Mac OS X 10_13_6) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.5735.90 Safari/537.36")
        self.options.add_argument("--window-size=1920,1080")
        self.options.add_argument("--disable-extensions")
        self.options.add_argument("--proxy-server='direct://'")
        self.options.add_argument("--proxy-bypass-list=*")
        self.options.add_argument("--start-maximized")
        
        # Check if running in Docker (no DISPLAY variable typically means Docker/headless)
        is_docker = os.getenv('CHROME_BIN') is not None or os.getenv('CHROMEDRIVER_PATH') is not None
        if is_docker and self.debug:
            print("Warning: Running in Docker environment. Browser UI may not be available. Forcing headless mode.")
            self.debug = False
        
        if self.debug == False:
            self.options.add_argument('--headless')
            self.options.add_argument('--headless=new')  # Use new headless mode for better compatibility
        
        self.options.add_argument('--disable-gpu')
        self.options.add_argument('--disable-dev-shm-usage')
        self.options.add_argument('--no-sandbox')
        self.options.add_argument('--ignore-certificate-errors')
        # Additional options for Docker/headless mode
        self.options.add_argument('--disable-blink-features=AutomationControlled')
        self.options.add_experimental_option("excludeSwitches", ["enable-automation", "enable-logging"])
        self.options.add_experimental_option('useAutomationExtension', False)

    def generate_random_id(self):
        # uuid4().int is 128-bit; PostgreSQL BIGINT fits 63-bit positive values
        return uuid.uuid4().int & 0x7FFFFFFFFFFFFFFF
    
    def retrieve_dom_and_generate_file(self, target_url):
        """
        Retrieves the DOM tree from a target URL and generates a file to store the code structure.
        :param target_url: The URL to retrieve the DOM from
        :return: None
        """
        try:
            print("-------------------------------------------------------------------------")
            print(f"Retrieving DOM tree from {target_url} and generating file...")
            print("-------------------------------------------------------------------------")
            # Load the URL
            self.load_url(target_url)
            # Wait for the page to load completely
            WebDriverWait(self.driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
            # Get the page source (DOM tree)
            page_source = self.driver.page_source
            # Generate a timestamp for the filename
            from datetime import datetime
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            # Create filename with timestamp
            filename = f"dom_tree_{timestamp}.html"
            # Write the DOM tree to a file
            with open(filename, 'w', encoding='utf-8') as file:
                file.write(page_source)
            print(f"DOM tree successfully saved to file: {filename}")
            print(f"File size: {len(page_source)} characters")
            # Also save a prettified version using BeautifulSoup if available
            try:
                from bs4 import BeautifulSoup
                soup = BeautifulSoup(page_source, 'html.parser')
                prettified_filename = f"dom_tree_prettified_{timestamp}.html"
                with open(prettified_filename, 'w', encoding='utf-8') as file:
                    file.write(soup.prettify())
                print(f"Prettified DOM tree saved to file: {prettified_filename}")
            except ImportError:
                print("BeautifulSoup not available, skipping prettified version")
            print("-------------------------------------------------------------------------")
        except Exception as e:
            print(f"Error retrieving DOM tree: {e}")
            print("-------------------------------------------------------------------------")

    def isAtBottom(self, max_scrolls=1):
        print("Scrolling down...", end='', flush=True)
        driver = self.driver
        lastHeight = driver.execute_script("return document.documentElement.scrollHeight")
        while True:
            driver.execute_script("var scrollingElement = (document.scrollingElement || document.body);scrollingElement.scrollTop = scrollingElement.scrollHeight;")
            driver.execute_script("window.scrollTo(0, document.documentElement.scrollHeight);")
            time.sleep(3)
            height = driver.execute_script("return document.documentElement.scrollHeight")
            driver.execute_script("window.scrollTo(0, " + str(height) + ");")
            if lastHeight == height:
                print("finished")
                break
            lastHeight = height
            #max_scrolls -= 1
    
    def element_exists(self, by, value):
        """
        Checks if an HTML element exists.
        
        :param by: Locator strategy (e.g., By.XPATH, By.ID, etc.)
        :param value: The value of the locator (e.g., XPath string, ID value, etc.)
        :return: True if the element exists, False otherwise
        """
        try:
            self.driver.find_element(by, value)
            return True
        except NoSuchElementException:
            print("no such element exists: ", value)
            return False
        
    def checkIfHTMLElementIsLoaded(self, html_element):
        html_filter_by = By.XPATH
        try:
            wait.until(EC.presence_of_element_located((html_filter_by, html_element)))
        except Exception as e:
            print("error while loading html element:", e)
        return element_exists(html_filter_by, html_element)
        
    def getText(self, el):
        try:
            return el.text
        except Exception as e:
            print("error during getting text from html el:", e)

    def extractToText(self, html_element):
        print("extractToText reached")
        wait = WebDriverWait(self.driver, 3.0)
        extracted_text = ''
        try:
            wait.until(EC.presence_of_element_located((By.XPATH, html_element)))
            print("html_element: ", html_element)
            extracted_text = self.driver.find_element(By.XPATH, html_element).text
            print("extracted_text: ", extracted_text)
        except Exception as e:
            print("error during extract_to_text trying alt:", e)
            try:
                html_element = '/html/body/div[1]/div/div/div[1]/div/div[3]/div[1]/div/div[1]/div/div/div/div[1]/h1'
                wait.until(EC.presence_of_element_located((By.XPATH, html_element)))
                print("html_element: ", html_element)
                extracted_text = self.driver.find_element(By.XPATH, html_element).text
                print("extracted_text: ", extracted_text)
            except Exception as e:
                print("error during extract_to_text 2nd:", e)
        return extracted_text
    
    def printXPathAndClassV2(self, exp_attr, exp_xpath, el):
        # Get the XPath of the element
        element_xpath = el.get_attribute("xpath")
        print(f"Expected xpath: {exp_xpath}. Current xpath: {element_xpath}")

        # Get the class attribute of the element
        element_class = el.get_attribute("class")
        print(f"Expected attr: {exp_attr}. Current attr: {element_class}")
    
    def getAttrValue(self, attr_value, el):
        try:
            return el.get_attribute(attr_value)
        except Exception as e:
            print("error getting html el attrb: ", e)
        
    def printChildHTMLElements(self, el):
        # Find and print all child elements of the <li> element
        child_elements = el.find_elements(By.XPATH, "./*")
        for child_element in child_elements:
            print(child_element.get_attribute("outerHTML"))
    
    def getHref(self, el):
        # Assuming div_child_4 is the h3 element
        # Find the nested anchor element within the h3 element using XPath
        anchor_element = el.find_element(By.XPATH, ".//a")

        # Retrieve the href attribute value from the anchor element
        href_value = anchor_element.get_attribute("href")

        # Print or use the href value as needed
        print("Href:", href_value)
        return href_value
    
    def isNewsInTimeframe(self, news_timestamp, timeframes):
        """
        Checks if the news timestamp falls within the specified timeframe.

        :param news_timestamp: The timestamp string (e.g., "2 days ago", "yesterday").
        :param timeframe: A list of strings where each element represents the timeframe to match (e.g., "yesterday", "2 days").
        :return: True if the news timestamp matches the timeframe, False otherwise.
        """
        # Check if the timeframe exists in the news timestamp
        for timeframe in timeframes:
            if timeframe in news_timestamp:
                print(f"News published '{timeframe}' reached.")
                return True
        return False

    def cancelPageReload(self):
        """
        Cancels a stuck or continuous page reload using JavaScript.
        """
        try:
            self.driver.execute_script("window.stop();")
            print("[INFO] Page reload was stopped via JavaScript.")
        except Exception as e:
            print("[ERROR] Failed to stop the page reload.")
            print(str(e))

    ####################################################################################################
    def getMainHeadline(self, html_element):
        return self.extractToText(html_element)
    
    def setUlNewsDivContainer(self):
        return '/html/body/div[1]/div/div/div[1]/div/div[3]/div[1]/div/div[3]/div/ul/li[1]/div/div/div[2]'
    
    def setUlNewsSourceAndTimestamp(self):
        return '/html/body/div[1]/div/div/div[1]/div/div[3]/div[1]/div/div[3]/div/ul/li[1]/div/div/div[2]/div[2]'

    def selectHTMLElement(self, root_el, query_by, query_html_el):
        return root_el.find_element(query_by, query_html_el)
    
    def selectHTMLElements(self, root_el, query_by, query_html_el):
        return root_el.find_elements(query_by, query_html_el)
    
    def extractUlNewsSourceAndTimestamp(self, el):
        query_by = By.TAG_NAME
        query_html_el = 'span'
        news_source = ''
        calculated_datetime = ''
        calculated_timestamp = ''
        els = self.selectHTMLElements(el, query_by, query_html_el)
        for i, el in enumerate(els):
            if i == 0:
                news_source = self.getText(el)
            elif i == 1:
                news_timestamp = self.getText(el)
                #calculated_datetime, calculated_timestamp = calculate_datetime_from_ago_string(news_timestamp)
        #print('news_source: ', news_source, ', calculated_datetime: ', calculated_datetime, ', calculated_timestamp', calculated_timestamp)
        return news_source
      
    def extractNewsHeaderSourceAndTimestamp(self, el):
        """
        Extracts the news source and timestamp from the specified HTML element.
        :param el: The parent HTML element containing the news source and timestamp.
        :return: A tuple (news_source, news_timestamp)
        """
        # Extract the text content of the element
        text = self.getText(el).strip()

        # Split the content based on the separator (e.g., "•")
        parts = text.split('•')
        news_source = None
        news_timestamp = None
        if len(parts) == 2:
            news_source = parts[0].strip()
            news_timestamp = parts[1].strip()

        return news_source,news_timestamp
    
    def extractFullNewsSourceAndTimestamp(self, el):
        query_by = By.TAG_NAME
        query_html_el = 'time'
        class_attribute = 'datetime'
        news_source = ''
        news_datetime = ''
        el = self.selectHTMLElement(el, query_by, query_html_el)
        print('full news_datetime: ', el.get_attribute(class_attribute))
        text = el.get_attribute(class_attribute)
        return text
    
    def extractFullNewsMinsRead(self, el):
        query_by = By.TAG_NAME
        query_html_el = 'span'
        mins_read = ''
        el = self.selectHTMLElement(el, query_by, query_html_el)
        mins_read = self.getText(el)
        print('full mins read: ', mins_read)
        return mins_read
    
    def clickOnReadMoreArticle(self):
        try:
            # Find all elements with the class 'caas-readmore'
            readmore_buttons = self.driver.find_elements(By.CLASS_NAME, "readmore-button")

            # Check if any buttons were found
            if readmore_buttons:
                # Click on the first button found
                readmore_buttons[0].click()
            else:
                print("No buttons with class 'caas-readmore' found")
        except Exception as e:
            print('Error on clickOnReadMoreArticle method: ', e)
    
    def selectSectionNestedHTMLElements(self, el, reach_time_limit, attempt=1, error_log=None, resume_ctx=None):
        if reach_time_limit[0]:
            return
        import time
        if error_log is None:
            error_log = []
        print("-------------------------------------------------------------------------")
        print("on selectSectionNestedHTMLElements method")
        print("-------------------------------------------------------------------------")
        # html 'el' should be the section item (section) html element
        query_by = By.XPATH
        query_html_el = './*'
        clickable_el = ''
        expected_html_attr = ''
        expected_html_xpath = ''
        attrValue=''
        attr='class'
        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,el)
        timeframe = ['1d']
        news_item = {}
        news_item["id"] = self.generate_random_id()
        news_item['summary'] = ""
        news_item['content'] = pd.NA
        news_item['author'] = pd.NA
        news_item['minsread'] = ""
        news_item['datetime'] = pd.NA
        def retry_or_print(err, context):
            msg = f"{context}: {err}"
            error_log.append(msg)
            if attempt < 3:
                print(f"Attempt {attempt}/3 failed in selectSectionNestedHTMLElements. Reloading page and retrying...")
                print("Error:", msg)
                if resume_ctx is not None:
                    resume_ctx["needs_rehydrate"] = True
                    resume_ctx["pending_li_attempt"] = attempt + 1
                    print(
                        "Page refreshed; newsHeaderScraper will call scrollNewsStreamSectionIntoViewAndWaitForUl() "
                        f"and resume stream list at index {resume_ctx.get('next_li_index', 0)}."
                    )
                try:
                    self.driver.refresh()
                except Exception as refresh_err:
                    refresh_msg = f"Refresh failed: {refresh_err}"
                    error_log.append(refresh_msg)
                    print(refresh_msg)
                time.sleep(1)
            else:
                print("selectSectionNestedHTMLElements failed after 3 attempts. Errors:")
                for idx, saved_error in enumerate(error_log, start=1):
                    print(f"  {idx}. {saved_error}")
                if resume_ctx is not None:
                    resume_ctx["give_up_current_li"] = True
        try:
            if 'list-item neo-pt-5 yf-12gc3ad' in self.getAttrValue(attr,el):
                try:
                    expected_html_attr = 'li, class=list-item neo-pt-5 yf-12gc3ad'
                    self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,el)
                    div_parent_0 = self.selectHTMLElements(el, query_by, query_html_el)
                    for a_0, div_chlds0 in enumerate(div_parent_0):
                        print("section, div_chlds0 loop 'a_0': ", a_0)
                        expected_html_attr = 'section, class=story-item horizontal-story variant-stream yf-axs8t2'
                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_chlds0)
                        try: 
                            div_parent = self.selectHTMLElements(div_chlds0, query_by, query_html_el)
                            for a, div_chlds1 in enumerate(div_parent):
                                print("div, div_chlds1 loop 'a': ", a)
                                expected_html_attr='div, class="content yf-1mnyyqc"'
                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_chlds1)
                                if 'content yf-1mnyyqc' in self.getAttrValue(attr,div_chlds1):
                                    div_chlds2 = self.selectHTMLElements(div_chlds1, query_by, query_html_el)
                                    try:
                                        for b, div_child2 in enumerate(div_chlds2):
                                            print("div_child2 loop b: ", b)
                                            if b == 0:
                                                expected_html_attr='div, class="title-row yf-axs8t2 has-clamp"'
                                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child2)
                                                div_chlds3 = self.selectHTMLElements(div_child2, query_by, query_html_el)
                                                for c_0, div_child3 in enumerate(div_chlds3):
                                                    print("div_child3 loop c_0: ", c_0)
                                                    if c_0 == 0:
                                                        expected_html_attr='a, class="hyperlink-wrapper variant-default level subtle yf-la14811z"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child3)
                                                        news_href = div_child3.get_attribute("href")
                                                        news_item['href'] = news_href
                                                        print("href: ", div_child3.get_attribute("href"))
                                                        div_chlds4 = self.selectHTMLElements(div_child3, query_by, query_html_el)
                                                        for c_0_n_0, div_child4 in enumerate(div_chlds4):
                                                            print("h3 div_child4 loop c_0_n_0: ", c_0_n_0)
                                                            if c_0_n_0 == 0:
                                                                expected_html_attr='h3, class="text neo-font-label-x1-emphasis yf-20ihx"'
                                                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child4)
                                                                print("headline: ",self.getText(div_child4))
                                                                news_item['headline'] = self.getText(div_child4)
                                            elif b == 1:
                                                expected_html_attr='div, class="byline yf-axs8t2"'
                                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child2)
                                                b_1_div_chlds3 = self.selectHTMLElements(div_child2, query_by, query_html_el)
                                                for c_1, b_1_div_child3 in enumerate(b_1_div_chlds3):
                                                    if c_1 == 0:
                                                        expected_html_attr='span, class="publishing yf-1weyqlp"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,b_1_div_child3)
                                                        news_source, news_timestamp = self.extractNewsHeaderSourceAndTimestamp(b_1_div_child3)
                                                        if self.isNewsInTimeframe(news_timestamp,timeframe) == True:
                                                            reach_time_limit[0] = True
                                                            return
                                                        news_item['source'] = news_source
                                                        print("news source: ", news_source)
                                                        print("news header timestamp: ", news_timestamp)
                                                        self.df_store.update_data_frame(news_item)                              
                                    except Exception as e:
                                        retry_or_print(e, "after defining div_chlds2")
                                        return
                        except Exception as e:
                            retry_or_print(e, "error on 'section, class=story-item horizontal-story variant-stream yf-axs8t2' block")
                            return
                                
                except Exception as e:
                    retry_or_print(e, "error while checking 'list-item neo-pt-5 yf-12gc3ad' block nested elements")
                    return
        except Exception as e:
                    retry_or_print(e, "error on 'list-item neo-pt-5 yf-12gc3ad' block")
                    return
        finally:
            print("-------------------------------------------------------------------------------------------------------------------------")
            print(f"finished scraping news header article section with attribute '{self.getAttrValue(attr,el)}'.")
            print("-------------------------------------------------------------------------------------------------------------------------")
    def getUlNewsCategory(self, el):
        pass

    def getUlNewsSource(self, el):
        pass
    
    def getUlNewsTimestamp(self, el):
        pass
    
    def getUlNewsSource(self, el):
        pass
    
    def selectFullNewsElements(self, el):
        print("-------------------------------------------------------------------------")
        print('on selectFullNewsElements() method')
        print("-------------------------------------------------------------------------")
        # html 'el' should be a list of div elements from the class 'caas-content-wrapper'
        # el = '/html/body/div[4]/div/div/div/div[2]/div[1]/div/div[1]/div/article/div/div/div/div/div/div[2]'
        query_by = By.XPATH
        query_html_el = './*'
        news_item = {}
        expected_html_attr = ''
        expected_html_xpath = ''
        attr = 'class'
        try:
            div_parent = el
            print("full_div_parent loop")
            for div_child in div_parent:
                expected_html_attr = 'div, class="article-wrap no-bb"'
                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child)
                div_chlds2 = self.selectHTMLElements(div_child, query_by, query_html_el)
                for i, div_child_2 in enumerate(div_chlds2):
                    print("full_div_chlds2 loop, i=", i)
                    try:
                        if "byline" in self.getAttrValue(attr,div_child_2):
                            expected_html_attr='div, class="byline yf-77f3vn"'
                            self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_2)
                            div_chlds3 = self.selectHTMLElements(div_child_2, query_by, query_html_el)
                            for j, div_child_3 in enumerate(div_chlds3): 
                                print("full_div_chlds3 loop, j=", j)
                                if j == 0:
                                    expected_html_attr='div, class="byline-attr yf-77f3vn"'
                                    self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_3)
                                    div_chlds4 = self.selectHTMLElements(div_child_3, query_by, query_html_el)
                                    for l, div_child_4 in enumerate(div_chlds4): 
                                        print("full_div_chlds4 loop, l=", l)
                                        if len(div_chlds4) == 1:
                                            if l == 0:
                                                expected_html_attr='div, class=""' # <div></div>
                                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_4)
                                                div_chlds5 = self.selectHTMLElements(div_child_4, query_by, query_html_el)
                                                for m, div_child_5 in enumerate(div_chlds5): 
                                                    print("full_div_chlds5 loop, m=", m)
                                                    if m == 0:
                                                        expected_html_attr='div, class="byline-attr-author yf-1k5w6kz"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_5)
                                                        print("full news author: ",self.getText(div_child_5))
                                                        news_item['author'] = self.getText(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "author"] = news_item['author']
                                                    elif m == 1: 
                                                        expected_html_attr='div, class="byline-attr-time-style"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_5)
                                                        print("full news published date: ",self.extractFullNewsSourceAndTimestamp(div_child_5))
                                                        news_item['datetime'] = self.extractFullNewsSourceAndTimestamp(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "datetime"] = news_item['datetime']
                                                        print("full news minsread: ",self.extractFullNewsMinsRead(div_child_5))
                                                        news_item['minsread']  = self.extractFullNewsMinsRead(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "minsread"] = news_item['minsread']
                                        elif len(div_chlds4) == 2:           
                                            if l == 1:
                                                expected_html_attr='div, class=""' # <div></div>
                                                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_4)
                                                div_chlds5 = self.selectHTMLElements(div_child_4, query_by, query_html_el)
                                                for m, div_child_5 in enumerate(div_chlds5): 
                                                    print("full_div_chlds5 loop, m=", m)
                                                    if m == 0:
                                                        expected_html_attr='div, class="byline-attr-time-style"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_5)
                                                        print("full news author: ",self.getText(div_child_5))
                                                        news_item['author'] = self.getText(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "author"] = news_item['author']
                                                    elif m == 1: 
                                                        expected_html_attr='div, class="byline-attr-time-style"'
                                                        self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_5)
                                                        print("full news published date: ",self.extractFullNewsSourceAndTimestamp(div_child_5))
                                                        news_item['datetime'] = self.extractFullNewsSourceAndTimestamp(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "datetime"] = news_item['datetime']
                                                        print("full news minsread: ",self.extractFullNewsMinsRead(div_child_5))
                                                        news_item['minsread']  = self.extractFullNewsMinsRead(div_child_5)
                                                        self.df_store.data_frame.at[self.df_store.row_index, "minsread"] = news_item['minsread']
                                                        
                        elif "body-wrap" in self.getAttrValue(attr,div_child_2):
                            print("div_child_2 loop, i=", i)
                            expected_html_attr='div, class="body-wrap yf-40hgrf"'
                            self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,div_child_2)
                            news_item['content'] = self.getText(div_child_2)
                            print("full news content: ", self.getText(div_child_2))
                            self.df_store.data_frame.at[self.df_store.row_index, "content"] = news_item['content']

                    except Exception as e:
                                print('While defining div_p1 on Full news, error is: ', e)
        except Exception as e:
            print('While defining div_parent, error is: ', e)
    
    def viewFullNewsArticle(self):
        #el.click()
        query_html_el='article-wrap'
        query_by=By.CLASS_NAME
        wait = WebDriverWait(self.driver, 1)
        try:
            el = wait.until(EC.presence_of_element_located((query_by, query_html_el)))
            expected_html_attr='div, class="article-wrap no-bb"'
            expected_html_xpath=''
            self.printXPathAndClassV2(expected_html_attr, expected_html_xpath,el)
            self.clickOnReadMoreArticle()
            query_by = By.XPATH
            query_html_el = '//article[@data-testid="article-content-wrapper"]'
            el = self.selectHTMLElements(el, query_by, query_html_el)
            self.selectFullNewsElements(el)
            content_value = self.df_store.data_frame.at[self.df_store.row_index, "content"]
            return pd.notna(content_value) and str(content_value).strip() and str(content_value).strip() != "nan"
        except Exception as e:
                print("error couldn't load full news article:", e)
                return False
                
    def selectUnorderList(self, html_element=''):
        html_element='/html/body/div[1]/div/div/div[1]/div/div[3]/div[1]/div/div[3]/div/div/div/ul'
        show_text = False
        print("In selectUnorderList method")
        print("html element:", html_element)
        ul_element = self.driver.find_element(By.XPATH, html_element)
        list_items = ul_element.find_elements(By.TAG_NAME, 'li')
        # Loop through each list item
        for li in list_items:
            if show_text == True:
                # Extract and print the text content of each list item
                li_text = li.text
                print("li text element: ", li_text)
            self.selectLiNewsElements(li)
    
    def waitForNewsFeedReady(self, timeout=20, min_items=3):
        """Wait until the UL feed has real list items and skeleton placeholders are gone."""
        wait = WebDriverWait(self.driver, timeout)

        def _feed_ready(driver):
            ul = driver.find_element(By.XPATH, UL_LIST_yf_12gc3ad)
            items = ul.find_elements(By.XPATH, "./li")
            if len(items) < min_items:
                return False

            # Yahoo rotates classes; detect common skeleton markers by class token.
            skeleton_nodes = ul.find_elements(
                By.XPATH,
                ".//*[contains(@class,'skeleton') or contains(@class,'shimmer') or contains(@class,'placeholder')]"
            )
            return ul if len(skeleton_nodes) == 0 else False

        return wait.until(_feed_ready)

    def scrollToModularBlockNewsStreamAndPrepareFeed(self):
        """Scroll to #modular-block-news-stream, wait for UL_LIST_yf_12gc3ad, then isAtBottom(). Returns True on success."""
        try:
            wait_secs = PAGE_LOAD_TIMEOUT
        except NameError:
            wait_secs = 30
        driver = self.driver
        print("Scrolling into view: #modular-block-news-stream ...")
        try:
            anchor = driver.find_element(By.ID, MODULAR_BLOCK_NEWS_STREAM_ID)
        except NoSuchElementException:
            print("Could not find #modular-block-news-stream; news stream section missing.")
            return False
        driver.execute_script("arguments[0].scrollIntoView({block: 'start'});", anchor)
        time.sleep(0.5)
        print(f"Waiting for news list (UL_LIST_yf_12gc3ad), up to {wait_secs}s ...")
        try:
            WebDriverWait(driver, wait_secs).until(
                EC.presence_of_element_located((By.XPATH, UL_LIST_yf_12gc3ad))
            )
        except TimeoutException:
            print("Timed out: UL_LIST_yf_12gc3ad did not appear.")
            return False
        print("News list UL is present. Scrolling to bottom to hydrate stream (isAtBottom) ...")
        self.isAtBottom()
        print("News stream section is ready; continuing with header scrape.")
        return True

    def scrollNewsStreamSectionIntoViewAndWaitForUl(self):
        """Scroll #modular-block-news-stream into view, wait for UL_LIST_yf_12gc3ad, hydrate stream. Call again after any full page refresh."""
        return self.scrollToModularBlockNewsStreamAndPrepareFeed()

    def newsHeaderScraper(self):
        print("In newsHeaderScraper method")
        import time
        # list (mutable) so selectSectionNestedHTMLElements can set the flag for the caller
        reach_time_limit = [False]
        max_refresh_retries = 3
        refresh_attempt = 0
        resume_ctx = {
            "next_li_index": 0,
            "needs_rehydrate": False,
            "pending_li_attempt": 1,
            "give_up_current_li": False,
        }

        while not reach_time_limit[0]:
            if not self.scrollNewsStreamSectionIntoViewAndWaitForUl():
                refresh_attempt += 1
                print(f"newsHeaderScraper: news stream preparation failed ({refresh_attempt}/{max_refresh_retries}). Retrying...")
                if refresh_attempt > max_refresh_retries:
                    print("newsHeaderScraper: reached max refresh retries. Stopping.")
                    break
                print(f"Refreshing page (attempt {refresh_attempt})...")
                self.driver.refresh()
                time.sleep(2)
                continue

            try:
                ul_parent_element = self.driver.find_element(By.XPATH, UL_LIST_yf_12gc3ad)

                expected_html_attr='ul class="list"'
                expected_html_xpath = '/html/body/div[1]/div[4]/main/section/section/section/section/div/div/section[10]/section/div/div/div/div/ul'
                self.printXPathAndClassV2(expected_html_attr, expected_html_xpath, ul_parent_element)

                ul_item = ul_parent_element.find_elements(By.XPATH, "./li")
                if not ul_item:
                    ul_item = ul_parent_element.find_elements(By.XPATH, ".//li")

                if not ul_item:
                    ul_item = ul_parent_element.find_elements(By.XPATH, "./li")

                if not ul_item:
                    raise Exception("No <li> items found in stream root after loader reached ready state")

                start = resume_ctx["next_li_index"]
                if start >= len(ul_item):
                    print(f"newsHeaderScraper: resume index {start} is past end of list ({len(ul_item)} items); stopping.")
                    break

                print(f"Processing stream <li> items from index {start} (total {len(ul_item)}).")

                for idx in range(start, len(ul_item)):
                    if reach_time_limit[0]:
                        break
                    li = ul_item[idx]
                    resume_ctx["next_li_index"] = idx
                    resume_ctx["needs_rehydrate"] = False
                    resume_ctx["give_up_current_li"] = False
                    att = resume_ctx.get("pending_li_attempt", 1)
                    self.selectSectionNestedHTMLElements(
                        li, reach_time_limit, attempt=att, resume_ctx=resume_ctx
                    )
                    if resume_ctx.get("needs_rehydrate"):
                        print(
                            "newsHeaderScraper: re-fetching stream after refresh; "
                            "scrollNewsStreamSectionIntoViewAndWaitForUl() runs at start of next loop iteration."
                        )
                        break
                    if resume_ctx.get("give_up_current_li"):
                        resume_ctx["pending_li_attempt"] = 1
                        resume_ctx["next_li_index"] = idx + 1
                        continue
                    resume_ctx["pending_li_attempt"] = 1
                    resume_ctx["next_li_index"] = idx + 1

                if resume_ctx.get("needs_rehydrate"):
                    continue

                if reach_time_limit[0]:
                    refresh_attempt = 0
                    break

                if resume_ctx["next_li_index"] >= len(ul_item):
                    print("newsHeaderScraper: finished all stream items in this pass.")
                    break

                # Reset refresh attempts after a successful hydration/iteration cycle.
                refresh_attempt = 0

            except Exception as e:
                refresh_attempt += 1
                print(f"newsHeaderScraper timeout/error attempt {refresh_attempt}/{max_refresh_retries}: {e}")
                # #region agent log
                self._debug_log(
                    run_id="run-1",
                    hypothesis_id="H5",
                    location="NewsCollector-Staging.ipynb:newsHeaderScraper:exception",
                    message="newsHeaderScraper exception",
                    data={"refresh_attempt": refresh_attempt, "max_refresh_retries": max_refresh_retries, "error": str(e)}
                )
                # #endregion

                if refresh_attempt > max_refresh_retries:
                    print("newsHeaderScraper: reached max refresh retries. Stopping.")
                    break

                print(f"PAGE_LOAD_TIMEOUT reached or feed not ready. Refreshing page (attempt {refresh_attempt})...")
                self.driver.refresh()
                time.sleep(2)
                print(
                    "newsHeaderScraper: resuming from stream index",
                    resume_ctx["next_li_index"],
                    "; next loop iteration will call scrollNewsStreamSectionIntoViewAndWaitForUl() before re-querying the list.",
                )
                continue
            
    def clickNextPage(self):
        try:
            import time
            el = WebDriverWait(self.driver, PAGE_LOAD_TIMEOUT).until(
                EC.element_to_be_clickable((By.XPATH, A_PRIMARY_LINK_FIN_SIZE_SMALL_yf_1119g04z))
            )
            el.click()
            time.sleep(5)
            print("Clicked next page (A_PRIMARY_LINK_FIN_SIZE_SMALL_yf_1119g04z)")
            print("Current URL:", self.driver.current_url)
        except Exception as e:
            print("clickNextPage:", e)

    def loadFullNewsStory(self):
        # Clicks on the button to show the full news
        el = '/html/body/div[3]/div[1]/div/main/div[1]/div/div/div/div/article/div/div/div/div/div/div[2]/div[3]/div/button'
        body_class = 'caas-body'

    def setFullNewContainer(self):
        return '/html/body/div[4]/div/div/div/div[2]/div[1]/div/div[1]/div/article/div/div/div/div/div/div[2]'
    
    def getNewsContentDivContainer(self):
        # div element that contains full news body
        return '/html/body/div[3]/div[1]/div/main/div[1]/div/div/div/div/article/div/div/div/div/div/div[2]/div[3]'

    def getTextContent(self, el):
        div_element = self.driver.find_element(By.XPATH, el)

        # Locate all <p> and <h2> elements within the div
        p_and_h2_elements = div_element.find_elements(By.XPATH, './/p | .//h2')

        # Loop through each <p> and <h2> element
        for element in p_and_h2_elements:
            # Extract and print the text content of each element
            element_text = element.text
            print("Element Text:", element_text)
    
    def printDataframe(self):
        print("-------------------------------------------------------------------------")
        print("on printDataframe()")
        print("-------------------------------------------------------------------------")
        # Iterate through each row
        for index, row in self.df_store.data_frame.iterrows():
            print(row)
    
    def retrieveFullNewsArticle(self, start_from_index=None, skip_processed=True):
        """
        Retrieve full news article content for all articles in the dataframe.
        Can resume from a specific index and skip already processed articles.
        
        :param start_from_index: Index to start from (None = start from beginning)
        :param skip_processed: If True, skip articles that already have content
        :return: Dictionary with processing statistics
        """
        print("-------------------------------------------------------------------------")
        print("on retrieveFullNewsArticle()")
        print("-------------------------------------------------------------------------")
        
        # Use global timeout configuration
        try:
            article_timeout = ARTICLE_LOAD_TIMEOUT
            max_retries = MAX_RETRIES
        except NameError:
            # Fallback if variables not defined
            article_timeout = 20
            max_retries = 2
        
        # Statistics
        stats = {
            'total': 0,
            'processed': 0,
            'skipped': 0,
            'failed': 0,
            'already_processed': 0
        }
        
        # Check if dataframe exists and has data
        if self.df_store.data_frame is None or len(self.df_store.data_frame) == 0:
            print("⚠ No articles in dataframe to process.")
            return stats
        
        stats['total'] = len(self.df_store.data_frame)
        print(f"Total articles to process: {stats['total']}")
        
        # Determine starting index
        if start_from_index is None:
            start_from_index = 0
        
        print(f"Starting from index: {start_from_index}")
        if skip_processed:
            print("Skipping already processed articles (those with content)")
        
        # Iterate through each row starting from start_from_index
        for index in range(start_from_index, len(self.df_store.data_frame)):
            row = self.df_store.data_frame.iloc[index]
            href = row.get("href", "")
            headline = row.get("headline", f"Article {index}")
            
            print(f"\n[{index + 1}/{stats['total']}] Processing: {headline[:60]}...")
            print(f"  URL: {href}")
            
            # Check if article is already processed
            if skip_processed:
                content = row.get("content", pd.NA)
                # Check if content exists and is not NA/empty
                if pd.notna(content) and str(content).strip() and str(content).strip() != "nan":
                    print(f"  ✓ Already processed (has content). Skipping...")
                    stats['already_processed'] += 1
                    continue
            
            # Process the article with error handling
            try:
                # Use improved load_url with shorter timeout for individual articles
                success = self.load_url(href, timeout=article_timeout, max_retries=max_retries, verify_url=False)
                
                if not success:
                    print(f"  ⚠ Failed to load article after retries. Skipping...")
                    stats['failed'] += 1
                    continue
                
                # Set the row index for viewFullNewsArticle to update the correct row
                self.df_store.row_index = index
                
                # Extract full article content
                try:
                    article_loaded = self.viewFullNewsArticle()
                    if article_loaded:
                        stats['processed'] += 1
                        print(f"  ✓ Successfully processed article {index + 1}")
                    else:
                        print("  ⚠ Could not load full news article. Skipping success mark...")
                        stats['failed'] += 1
                        continue
                except Exception as e:
                    print(f"  ⚠ Error extracting content: {e}")
                    print(f"  Continuing to next article...")
                    stats['failed'] += 1
                    continue
                    
            except KeyboardInterrupt:
                print(f"\n⚠ Process interrupted by user at index {index}")
                print(f"  To resume, call: retrieveFullNewsArticle(start_from_index={index})")
                raise
            except Exception as e:
                print(f"  ⚠ Unexpected error processing article: {e}")
                print(f"  Continuing to next article...")
                stats['failed'] += 1
                continue
        
        # Print summary
        print("\n" + "=" * 70)
        print("Processing Summary:")
        print("=" * 70)
        print(f"  Total articles: {stats['total']}")
        print(f"  Successfully processed: {stats['processed']}")
        print(f"  Already processed (skipped): {stats['already_processed']}")
        print(f"  Failed: {stats['failed']}")
        print(f"  Remaining: {stats['total'] - stats['processed'] - stats['already_processed'] - stats['failed']}")
        print("=" * 70)
        
        return stats
    
    def get_last_processed_index(self):
        """
        Find the index of the last article that has been processed (has content).
        
        :return: Index of last processed article, or 0 if none processed
        """
        if self.df_store.data_frame is None or len(self.df_store.data_frame) == 0:
            return 0
        
        # Find the last index with content
        for index in range(len(self.df_store.data_frame) - 1, -1, -1):
            row = self.df_store.data_frame.iloc[index]
            content = row.get("content", pd.NA)
            if pd.notna(content) and str(content).strip() and str(content).strip() != "nan":
                return index + 1  # Return next index to process
        
        return 0
    
    def get_processing_status(self):
        """
        Get status of article processing.
        
        :return: Dictionary with processing statistics
        """
        if self.df_store.data_frame is None or len(self.df_store.data_frame) == 0:
            return {
                'total': 0,
                'processed': 0,
                'remaining': 0,
                'last_processed_index': 0
            }
        
        total = len(self.df_store.data_frame)
        processed = 0
        
        for index in range(total):
            row = self.df_store.data_frame.iloc[index]
            content = row.get("content", pd.NA)
            if pd.notna(content) and str(content).strip() and str(content).strip() != "nan":
                processed += 1
        
        last_index = self.get_last_processed_index()
        
        return {
            'total': total,
            'processed': processed,
            'remaining': total - processed,
            'last_processed_index': last_index,
            'percentage': (processed / total * 100) if total > 0 else 0
        }
    
    def retrieveFullNewsArticleAutoResume(self):
        """
        Automatically resume from the last processed article.
        This is a convenience wrapper that automatically finds where to resume.
        
        :return: Dictionary with processing statistics
        """
        status = self.get_processing_status()
        last_index = status['last_processed_index']
        
        print("=" * 70)
        print("Auto-Resume Full News Article Processing")
        print("=" * 70)
        print(f"Total articles: {status['total']}")
        print(f"Already processed: {status['processed']} ({status['percentage']:.1f}%)")
        print(f"Remaining: {status['remaining']}")
        print(f"Resuming from index: {last_index}")
        print("=" * 70)
        
        return self.retrieveFullNewsArticle(start_from_index=last_index, skip_processed=True)
    
    def storeToDatabase(self):
        print("-------------------------------------------------------------------------")
        print("storing to Database")
        print("-------------------------------------------------------------------------")
        pg_conn.reopen_connection()    
        header = self.df_store.data_frame_header
        for index, row in self.df_store.data_frame.iterrows():
            pg_conn.save_to_postgres(row, header)
            
    def getParagraphs(self, el):
        pass

    def getAuthor(self, element):
        pass

    def getHeadline(self, element):
        pass

    def getContent(self, element):
        pass

    def getSource(self, element):
        pass

    def getPublishedDate(self, element):
        pass

    def getCategory(self, element):
        pass

    def quit_driver(self):
        # Close the WebDriver
        self.driver.quit()
    
    def cancel_page_reload(self):
        """
        Cancels a stuck or continuous page reload using JavaScript.
        """
        try:
            self.driver.execute_script("window.stop();")
            print("[INFO] Page reload was stopped via JavaScript.")
        except Exception as e:
            print("[ERROR] Failed to stop the page reload.")
            print(str(e))

    def load_url(self, target_url, timeout=30, max_retries=3, verify_url=True):
        """
        Load a URL with timeout and retry logic.
        
        :param target_url: URL to load
        :param timeout: Maximum time to wait for page load in seconds
        :param max_retries: Maximum number of retry attempts
        :param verify_url: Whether to verify the final URL matches the target
        :return: True if loaded successfully, False otherwise
        """
        print("-------------------------------------------------------------------------")
        print(f"Loading target URL: {target_url}")
        print("-------------------------------------------------------------------------")
        
        for attempt in range(1, max_retries + 1):
            try:
                print(f"Attempt {attempt}/{max_retries}...", end='', flush=True)
                
                # Start loading the page
                self.driver.set_page_load_timeout(timeout)
                self.driver.get(target_url)
                
                # Wait for page to be in a loaded state (complete or interactive)
                try:
                    WebDriverWait(self.driver, timeout).until(
                        lambda d: d.execute_script("return document.readyState") in ["complete", "interactive"]
                    )
                except TimeoutException:
                    print(f" Page load timeout after {timeout}s. Cancelling...")
                    self.cancel_page_reload()
                    if attempt < max_retries:
                        print(f" Retrying in 2 seconds...")
                        time.sleep(2)
                        continue
                    else:
                        print(" Max retries reached. Moving on...")
                        return False
                
                # Verify URL if requested
                if verify_url:
                    current_url = self.driver.current_url
                    # Allow for URL variations (query params, redirects, etc.)
                    if target_url in current_url or current_url in target_url:
                        print(" ✓ Success")
                        print(f"  Final URL: {current_url}")
                        return True
                    elif "consent.yahoo.com" in current_url:
                        print(" ✓ Loaded (consent page detected)")
                        print(f"  Current URL: {current_url}")
                        # Try to handle consent page
                        try:
                            time.sleep(2)
                            # Try to find and click accept button if it exists
                            accept_selectors = [
                                "//button[contains(text(), 'Accept')]",
                                "//button[contains(text(), 'I agree')]",
                                "//button[@name='agree']"
                            ]
                            for selector in accept_selectors:
                                try:
                                    accept_btn = self.driver.find_element(By.XPATH, selector)
                                    accept_btn.click()
                                    time.sleep(2)
                                    break
                                except:
                                    continue
                        except:
                            pass
                        return True
                    else:
                        print(f" ⚠ URL mismatch (attempt {attempt}/{max_retries})")
                        print(f"  Expected: {target_url}")
                        print(f"  Got: {current_url}")
                        if attempt < max_retries:
                            print(f" Retrying in 2 seconds...")
                            time.sleep(2)
                            continue
                        else:
                            print(" Max retries reached. Continuing with current URL...")
                            return True  # Continue anyway
                else:
                    print(" ✓ Success (URL verification skipped)")
                    return True
                    
            except TimeoutException as toe:
                print(f" TimeoutException: {toe}")
                self.cancel_page_reload()
                if attempt < max_retries:
                    print(f" Retrying in 2 seconds...")
                    time.sleep(2)
                    continue
                else:
                    print(" Max retries reached.")
                    return False
                    
            except Exception as e:
                print(f" Error: {e}")
                if attempt < max_retries:
                    print(f" Retrying in 2 seconds...")
                    time.sleep(2)
                    continue
                else:
                    print(" Max retries reached.")
                    return False
        
        print("-------------------------------------------------------------------------")
        return False


    def selenium_scrapper(self, html_element):
        max_time_to_wait_html_element = 1.5
        skipScrollingDown = True
        # Use global timeout configuration
        try:
            page_load_timeout = PAGE_LOAD_TIMEOUT
            max_retries = MAX_RETRIES
        except NameError:
            # Fallback if variables not defined
            page_load_timeout = 30
            max_retries = 3
            
        try:
            WebDriverWait(self.driver,max_time_to_wait_html_element).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
            try:
                for topic in self.topics:
                    print(f'Topic: {topic}')
                    target_url = f"https://finance.yahoo.com/markets/{topic}/"
                    
                    # Use improved load_url with timeout and retry logic
                    success = self.load_url(target_url, timeout=page_load_timeout, max_retries=max_retries, verify_url=True)
                    
                    if not success:
                        print(f"⚠ Failed to load {target_url} after retries. Skipping this topic.")
                        continue
                    
                    # Additional verification - check if we're on the right page
                    current_url = self.driver.current_url
                    if target_url not in current_url and "consent.yahoo.com" not in current_url:
                        print(f"⚠ URL verification: Expected {target_url}, got {current_url}")
                        print("  Continuing anyway...")
                    
                    if skipScrollingDown == False:
                        self.isAtBottom()
                    self.newsHeaderScraper()
                    self.printDataframe()
                    # Use auto-resume to continue from last processed article if interrupted
                    self.retrieveFullNewsArticleAutoResume()
                    self.storeToDatabase()
                    print("scraping data task finished")
                    print("====================================================================")
                print("**All book data was scraped**")
            except NoSuchElementException as nse:
                print(nse)
                print("-----")
                print(str(nse))
                print("-----")
                print(nse.args)
                print("=====")
        except Exception as e:
            print(e)
            print("-----")
            print(str(e))
            print("-----")
            print(e.args)

        except TimeoutException as toe:
            print(toe)
            print("-----")
            print(str(toe))
            print("-----")
            print(toe.args)
        finally:
            '''
            if(DEBUG):
                delete_table("historical", conn)
            '''
            pg_conn.close_connection()
    
    def setMissingArticlesForFullNewsContent(self):
        # TODO get all dataframe articles with missing Full Content News data
        pass
    
    def selenium_scrapper_fullNewsData(self, html_element):
        # Initialize an empty list to store extracted data
        storeToDB = False
        try: 
            self.setMissingArticlesForFullNewsContent()
            self.retrieveFullNewsArticle()
            if storeToDB == True:
                self.storeToDatabase()
            #last_five_rows = df.tail(3)
            print("====================================================================")
            print("scraping data task finished")
            print("====================================================================")
            print("**All book data was scraped**")
        except NoSuchElementException as nse:
            print(nse)
            print("-----")
            print(str(nse))
            print("-----")
            print(nse.args)
            print("=====")
        except Exception as e:
            print(e)
            print("-----")
            print(str(e))
            print("-----")
            print(e.args)
        except TimeoutException as toe:
            print(toe)
            print("-----")
            print(str(toe))
            print("-----")
            print(toe.args)
        finally:
            if(DELETE_TABLE):
                pg_conn.delete_table()
            pg_conn.close_connection()
        
    def setTestDataForFullNewsContent(self):
        data = {
            "id": self.generate_random_id(),
            "source": "Test source",
            "headline": "Test headline",
            "href": "https://finance.yahoo.com/markets/crypto/articles/x-claims-cashtags-launch-drove-191022798.html",
            "summary": "Test summary",
            "content": pd.NA,
            "autor": pd.NA,
            "minsread": pd.NA,
            "datetime": pd.NA
        }
        self.df_store.create_data_frame
        self.df_store.update_data_frame(data)
    
    def test_HeaderNewsContent(self):
        max_time_to_wait_html_element = 1.5
        skipScrollingDown = False
        try:
            WebDriverWait(self.driver,max_time_to_wait_html_element).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
            try:
                for topic in self.topics:
                    print(f'Topic: {topic}')
                    target_url = f"https://finance.yahoo.com/topic/{topic}/"
                    self.load_url(target_url)
                    print("driver.current_url != target_url: ", self.driver.current_url != target_url)
                    while self.driver.current_url != target_url:
                        self.driver.get(target_url)
                        time.sleep(1) 
                    if skipScrollingDown == False:
                        self.isAtBottom()
                    self.newsHeaderScraper()
            except Exception as e:
                print('While testing header news, error is: ', e)
        except Exception as e:
            print('While waiting loading page body, error is: ', e)
        
    def test_fullNewContent(self):
        # Initialize an empty list to store extracted data
        storeToDB = False
        try: 
            self.setTestDataForFullNewsContent()
            self.retrieveFullNewsArticle()
            if storeToDB == True:
                self.storeToDatabase()
            #last_five_rows = df.tail(3)
            print("====================================================================")
            print("scraping data task finished")
            print("====================================================================")
            print("**All book data was scraped**")
        except NoSuchElementException as nse:
            print(nse)
            print("-----")
            print(str(nse))
            print("-----")
            print(nse.args)
            print("=====")
        except Exception as e:
            print(e)
            print("-----")
            print(str(e))
            print("-----")
            print(e.args)
        except TimeoutException as toe:
            print(toe)
            print("-----")
            print(str(toe))
            print("-----")
            print(toe.args)
        finally:
            if(DELETE_TABLE):
                pg_conn.delete_table()
            pg_conn.close_connection()
```

```python
from datetime import datetime

def parse_date(date_str):
    # Convert the month name to a numerical representation using a dictionary
    month_dict = {
        "Jan": "01",
        "Feb": "02",
        "Mar": "03",
        "Apr": "04",
        "May": "05",
        "Jun": "06",
        "Jul": "07",
        "Aug": "08",
        "Sep": "09",
        "Oct": "10",
        "Nov": "11",
        "Dec": "12",
    }

    date_str = date_str.replace(",", "")

    # Split the date string into month, day, and year
    month, day, year = date_str.split()

    # Get the numerical representation of the month from the dictionary
    month_number = month_dict[month]

    # Create a new date string in the format 'year-month-day' (e.g., '2023-08-01')
    formatted_date_str = f"{year}-{month_number}-{day}"

    # Parse the formatted date string to a datetime object
    parsed_date = datetime.strptime(formatted_date_str, "%Y-%m-%d")

    return parsed_date

def parse_row_data(row_data):
    try:
        date_format = '%Y-%m-%d'  # Format for parsing date strings

        # Remove commas from numeric values
        row_data = [item.replace(",", "") if isinstance(item, str) else item for item in row_data]

        # Parse elements at specific positions into desired data types
        row_data[0] = parse_date(row_data[0])
        row_data[1] = float(row_data[1])
        row_data[2] = float(row_data[2])
        row_data[3] = float(row_data[3])
        row_data[4] = float(row_data[4])
        row_data[5] = float(row_data[5])
        row_data[6] = int(row_data[6])
        return row_data
    except Exception as e:
        print("error during parsing data:", e, "row_data: ", row_data)
```
